# Exploratory Data Analysis

This notebook inspects the freMTPL2 frequency and severity datasets, shows descriptive statistics, missing-value analysis, and meaningful visualisations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_style('whitegrid')

freq = pd.read_csv(Path('../data/raw/freMTPL2freq.csv'))
sev = pd.read_csv(Path('../data/raw/freMTPL2sev.csv'))

print('Frequency dataset shape:', freq.shape)
print('Severity  dataset shape:', sev.shape)
```

## 1. Dataset Inspection

```python
print('=== Frequency Dataset ===')
print(freq.dtypes)
print()
print(freq.head())
print()
print('=== Severity Dataset ===')
print(sev.dtypes)
print()
print(sev.head())
```

## 2. Descriptive Statistics

```python
print('=== Frequency Descriptive Stats ===')
print(freq.describe())
print()
print('=== Severity Descriptive Stats ===')
print(sev.describe())
```

## 3. Missing-Value Analysis

```python
print('Missing values in frequency data:')
print(freq.isnull().sum())
print()
print('Missing values in severity data:')
print(sev.isnull().sum())
```

## 4. Claim Frequency Distribution

```python
fig, ax = plt.subplots(figsize=(8, 5))
claim_counts = freq['ClaimNb'].value_counts().sort_index()
ax.bar(claim_counts.index, claim_counts.values, color='steelblue')
ax.set_xlabel('Number of Claims')
ax.set_ylabel('Number of Policies')
ax.set_title('Claim Frequency Distribution')
ax.set_yscale('log')
plt.tight_layout()
plt.savefig('../reports/figures/claim_frequency_distribution.png', dpi=150)
plt.show()
```

## 5. Claim Amount Distribution

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(sev['ClaimAmount'], bins=100, color='coral', edgecolor='white', log=True)
ax.set_xlabel('Claim Amount (EUR)')
ax.set_ylabel('Count (log scale)')
ax.set_title('Claim Amount Distribution')
plt.tight_layout()
plt.savefig('../reports/figures/claim_amount_distribution.png', dpi=150)
plt.show()
```

## 6. Exposure Analysis

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(freq['Exposure'], bins=50, color='seagreen', edgecolor='white')
ax.set_xlabel('Exposure (years)')
ax.set_ylabel('Number of Policies')
ax.set_title('Exposure Distribution')
plt.tight_layout()
plt.savefig('../reports/figures/exposure_distribution.png', dpi=150)
plt.show()
```

## 7. Claim Frequency by Area

```python
freq['ClaimRate'] = freq['ClaimNb'] / freq['Exposure']
area_rates = freq.groupby('Area')['ClaimRate'].mean().sort_index()

fig, ax = plt.subplots(figsize=(8, 5))
area_rates.plot(kind='bar', ax=ax, color='mediumpurple', edgecolor='white')
ax.set_xlabel('Area')
ax.set_ylabel('Average Annualized Claim Rate')
ax.set_title('Average Claim Frequency by Area')
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.savefig('../reports/figures/claim_rate_by_area.png', dpi=150)
plt.show()
```
